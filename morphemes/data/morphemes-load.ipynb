## Get English morphemes in a sentence

```python
dfs = pd.read_excel("MorphoLEX_en.xlsx", sheet_name=None)

dfs.pop('Presentation')
dfs.pop('All prefixes')
dfs.pop('All suffixes')
dfs.pop('All roots')

for key in dfs:
    dfs[key] = dfs[key][["MorphoLexSegm", "PRS_signature", "Word"]]

dfs = list(dfs.values())
```

```python
morphemes = pd.concat(dfs).dropna().reset_index()
morphemes.pop("index")
morphemes["Word"] = morphemes["Word"].str.lower()

morphemes
```

```python
deleted_char = [",", ".", ";", "/", "!", "?", "(", ")", "\n", "-", ":"]

def clean_string(sentence:str):
    sentence = sentence.lower()
    for c in deleted_char:
        sentence = sentence.replace(c, " ")
    sentence.replace("  ", " ")
    return sentence

def split_sentence(sentence:str):
    sentence = clean_string(sentence)
    split = sentence.split(" ")
    split = [v for v in split if v not in ["", " "]]
    return split

def get_morphemes(sentence:str):
    split = split_sentence(sentence)
    #print(f"Sentence length: {len(split)} words")
    morph = []

    errors = []

    for w in split:
        res = morphemes.loc[morphemes["Word"]==w]
        if(len(res)!=0):
            morph.append(res.index[0])
        elif(len(w.split("'"))>1 and len(morphemes.loc[morphemes["Word"]==w.split("'")[0]])>0):
            morph.append(w.split("'")[0])
        else:
            errors.append(w)

    #print(f"> {len(morph)} recognized words: {errors}")

    return morph
```

## Rosetta Stone

```python
# DEBUG
rosetta_stone = [
    ("Hello, my name is", "Hello, my name is"),
    ("Hello, how are you ?", "Hello, how are you ?"),
    ("How is it going ?", "How is it going ?"),
    ("What do you do for living ?", "What do you do for living ?"),
    ("What is it ?", "What is it ?"),
    ("The book is green", "The book is green"),
    ("The sun is bright", "The sun is bright")
]
```

```python
rosetta_stone = [
    ("Hello, my name is", "こんにちは、私の名前は……です。"),
    ("Hello, how are you ?", "こんにちは お元気ですか ？"),
    ("How is it going ?", "調子はどうだい ？"),
    ("What do you do for living ?", "お仕事は何をされていますか？"),
    ("What is it ?", "それは何ですか ？"),
    ("The book is green", "その本は緑色です。"),
    ("The sun is bright", "太陽がまぶしいです。"),
    ("I do not know what it is", "それが何なのかは分かりません。"),
    ("I appreciate how you did that", "そうしてくださったことに感謝します。")
]

deleted_char += ["。", "、", "…", "？"]
```

<i>morphemes_stone</i> contient la décomposition en morphèmes de la phrase en Anglais et l'ensemble de caractères correspondants dans la langue étudiée.

<i>all_en_morphemes</i> contient tous les morphèmes apparaissant en Anglais et l'ensemble des phrases de la pierre de Rosette associées

```python
morphemes_stone = [(get_morphemes(an), clean_string(other)) for (an, other) in rosetta_stone]

all_en_morphemes = {}
all_trad_sentences = {}
for (lst, trad) in morphemes_stone:
    for morph in lst: 
        if morph in all_en_morphemes:
            all_en_morphemes[morph].append(set(lst))
            all_trad_sentences[morph].append(trad)
        else:
            all_en_morphemes[morph] = [set(lst)]
            all_trad_sentences[morph] = [trad]

#all_en_morphemes
```

```python
def intersect_sets(lst:list):
    assert(len(lst)>0)
    res = lst[0]
    for l in lst[1:]:
        res = res & l
    return res

identifiable = []
for morph in all_en_morphemes:
    inter = intersect_sets(all_en_morphemes[morph])
    assert(len(inter) > 0)
    if(len(inter)==1):
        print(f"Morpheme -{morphemes['Word'].iloc[morph]}- ({morph}) can be identified")
        identifiable.append(morph)
```

```python
def longest_common_substring_all(strings, case_sensitive=True):
    """
    POWERED BY AI xD
    Trouve la plus longue sous-chaine (contigue) commune a TOUTES les strings
    de la liste. Fonctionne bien meme avec un grand nombre de strings.
    Retourne "" si aucune sous-chaine non vide n'est commune a toutes.
    """
    if not strings:
        return ""
    if not case_sensitive:
        strings = [s.lower() for s in strings]

    n = len(strings)
    # On travaille sur la string la plus courte : la reponse ne peut pas
    # etre plus longue qu'elle.
    shortest = min(strings, key=len)
    others = [s for s in strings if s is not shortest] or strings[1:]

    MOD = (1 << 61) - 1
    BASE = 131

    def hashes_of_length(s, L):
        """Retourne un set des hachages de toutes les sous-chaines de longueur L de s."""
        if L == 0 or L > len(s):
            return set()
        h = 0
        p = 1
        for i in range(L):
            h = (h * BASE + ord(s[i])) % MOD
            if i < L - 1:
                p = (p * BASE) % MOD
        seen = {h}
        for i in range(L, len(s)):
            h = ((h - ord(s[i - L]) * p) * BASE + ord(s[i])) % MOD
            seen.add(h)
        return seen

    def common_hash_of_length(L):
        """Renvoie un hash commun a toutes les strings pour la longueur L, ou None."""
        candidate_hashes = hashes_of_length(shortest, L)
        if not candidate_hashes:
            return None
        for s in others:
            candidate_hashes &= hashes_of_length(s, L)
            if not candidate_hashes:
                return None
        return next(iter(candidate_hashes))

    def extract_substring_with_hash(L, target_hash):
        """Retrouve la sous-chaine reelle de longueur L dans `shortest` correspondant au hash."""
        for i in range(len(shortest) - L + 1):
            candidate = shortest[i:i + L]
            # verification directe pour eviter les faux positifs de hash
            h = 0
            for ch in candidate:
                h = (h * BASE + ord(ch)) % MOD
            if h == target_hash:
                return candidate
        return None

    lo, hi = 0, len(shortest)
    best_L = 0
    best_hash = None

    while lo <= hi:
        mid = (lo + hi) // 2
        h = common_hash_of_length(mid)
        if h is not None:
            best_L = mid
            best_hash = h
            lo = mid + 1
        else:
            hi = mid - 1

    if best_L == 0:
        return ""
    return str.strip(extract_substring_with_hash(best_L, best_hash))
```
